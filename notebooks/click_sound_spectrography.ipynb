```python
import numpy as np
import scipy.signal
```

```python
timebase = np.linspace(-1/int(250000*0.01),0.01, int(250000*0.01), endpoint=False)
signal = scipy.signal.square(2*np.pi*500*timebase)*2.5+2.5
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
plt.plot(timebase, signal)
```

```python
# create padding and concatenate as a sequence (np.concatenate expects a sequence)
pad_len = int(250000 * 0.02)
pad = np.linspace(0, 0.02, pad_len)
signal = np.concatenate([pad, signal, pad])
f, t, spek = scipy.signal.spectrogram(signal, fs=250000)
plt.pcolormesh(t,f,spek, shading='gouraud')
plt.ylim([0, 10000])
plt.show()
```
